```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
import pandas as pd

print("Libraries imported successfully!")
```

```python
# 1. what they are working for
job_description = """
We are looking for a highly skilled Python Developer with experience in web frameworks like Django or Flask.
The ideal candidate should have a strong understanding of RESTful APIs, database management (SQL),
and front-end technologies like React or Tailwind CSS. Experience with Git and cloud platforms is a plus.
"""

# Enter name of people
resumes = [
    {
        "name": "harsha",
        "text": "Full-stack developer with 3 years of experience. Proficient in Python, Django, and React. Built scalable REST APIs and managed PostgreSQL databases. Familiar with Tailwind CSS and version control using Git."
    },
    {
        "name": "thanmay",
        "text": "Graphic Designer with 5 years of experience using Adobe Creative Suite. Expert in Photoshop, Illustrator, and UI/UX design. Strong eye for detail and typography. Looking to transition into web design."
    },
    {
        "name": "prem",
        "text": "dental"
    }
]

print("Data loaded successfully!")
```

```python
model = SentenceTransformer('all-MiniLM-L6-v2')

job_embedding = model.encode([job_description])
resume_texts = [resume["text"] for resume in resumes]
resume_embeddings = model.encode(resume_texts)

similarity_scores = cosine_similarity(job_embedding, resume_embeddings)[0]

for i, resume in enumerate(resumes):
    resume["match_percentage"] = round(similarity_scores[i] * 100, 2)
```

```python
results_df = pd.DataFrame(resumes)

final_view = results_df[['name', 'match_percentage']].sort_values(by='match_percentage', ascending=False)

print("--- AI Resume Screening Results ---")
display(final_view)
```
